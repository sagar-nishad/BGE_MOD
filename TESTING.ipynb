```python
from datasets import load_dataset
from transformers import TrainingArguments
from FlagEmbedding.finetune.embedder.encoder_only.m3 import (
    EncoderOnlyEmbedderM3DataArguments,
    EncoderOnlyEmbedderM3TrainingArguments,
    EncoderOnlyEmbedderM3ModelArguments,
    EncoderOnlyEmbedderM3Runner,
)
# val_dataset = load_dataset("json" , data_files = "./ft_data/test.json")["train"]

from FlagEmbedding.abc.finetune.embedder.AbsDataset import AbsEmbedderTrainDataset
# Load model directly
from transformers import AutoTokenizer, AutoModelForMaskedLM

tokenizer = AutoTokenizer.from_pretrained("thebajajra/RexBERT-base")
model_args = EncoderOnlyEmbedderM3ModelArguments(model_name_or_path = "thebajajra/RexBERT-base" , tokenizer_name="thebajajra/RexBERT-base" , trust_remote_code = True)
data_args  = EncoderOnlyEmbedderM3DataArguments(train_data=["./ft_data/test.json"] , eval_data=["./ft_data/test.json"] , query_max_len=500 , passage_max_len= 500)


val_args  = EncoderOnlyEmbedderM3DataArguments(train_data=["./ft_data/test.json"] , query_max_len=500 , passage_max_len= 500)

val_dataset = AbsEmbedderTrainDataset( tokenizer = tokenizer , args = val_args)

# Instantiate the child class and pass ALL arguments
trainer_args = EncoderOnlyEmbedderM3TrainingArguments(
    
    # --- Arguments from base TrainingArguments ---
    output_dir="./models",
    overwrite_output_dir=True,
    learning_rate=5e-5,
    per_device_train_batch_size=8,
    per_device_eval_batch_size=8,
    num_train_epochs=1,
    seed=0,
    save_total_limit=1,
    do_train=True,
    do_eval=True,
    metric_for_best_model="eval_loss",
    greater_is_better=False,
    load_best_model_at_end=True,
    eval_strategy="steps",
    save_steps = 2  , 
    eval_steps=2,
    logging_steps=2,
    save_strategy="steps",

    # --- Arguments from AbsEmbedderTrainingArguments ---
    negatives_cross_device=False,
    temperature=0.02,
    sentence_pooling_method='cls',
    normalize_embeddings=True,
    # eval_dataset=val_dataset, 
    remove_unused_columns=False , 
    # --- Arguments from EncoderOnlyEmbedderM3TrainingArguments ---
    unified_finetuning=False,
    use_self_distill=False,
    fix_encoder=False
)


runner = EncoderOnlyEmbedderM3Runner(model_args=model_args, data_args=data_args, training_args=trainer_args  , eval_dataset= val_dataset )
runner.run()
runner.save_model("./ft_model/")
```
